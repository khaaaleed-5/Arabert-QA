```python
from datasets import Dataset

examples = []

for item in dataset:
    paragraphs = item["data"]["paragraphs"]

    for paragraph in paragraphs:
        context = paragraph["context"]

        for qa in paragraph["qas"]:
            if qa["is_impossible"]:
                continue

            if not qa["answers"]:
                continue

            answer = qa["answers"][0]

            examples.append({
                "id": str(qa["id"]),
                "question": qa["question"].strip(),
                "context": context,
                "answers": {
                    "text": [answer["text"]],
                    "answer_start": [answer["answer_start"]]
                }
            })

arabica_qa = Dataset.from_list(examples)

print(arabica_qa)
print(arabica_qa[0])
```

```python
def check_alignment(example):
    answer = example["answers"]["text"][0]
    start = example["answers"]["answer_start"][0]

    extracted = example["context"][start:start + len(answer)]

    return {
        "alignment_ok": extracted == answer
    }


results = arabica_qa.map(check_alignment)

print(results)
```

```python
from transformers import AutoTokenizer

MODEL_NAME = "aubmindlab/bert-base-arabertv02"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
```

```python
print("Vocab size:", tokenizer.vocab_size)
print("Model max length:", tokenizer.model_max_length)
```

```python
def tokenize_qa(examples):
    questions = [q.strip() for q in examples["question"]]
    contexts = examples["context"]

    tokenized = tokenizer(
        questions,
        contexts,
        max_length=384,
        truncation="only_second",
        stride=128,
        return_overflowing_tokens=True,
        return_offsets_mapping=True,
        padding="max_length",
    )

    sample_mapping = tokenized["overflow_to_sample_mapping"]

    start_positions = []
    end_positions = []

    for i, offsets in enumerate(tokenized["offset_mapping"]):

        sample_idx = sample_mapping[i]

        start_char = examples["answers"][sample_idx]["answer_start"][0]
        answer_text = examples["answers"][sample_idx]["text"][0]
        end_char = start_char + len(answer_text)

        sequence_ids = tokenized.sequence_ids(i)

        # Find context start
        context_start = sequence_ids.index(1)

        # Find context end
        context_end = len(sequence_ids) - 1

        while sequence_ids[context_end] != 1:
            context_end -= 1

        # Answer is not inside this feature
        if (
            offsets[context_start][0] > start_char
            or offsets[context_end][1] < end_char
        ):
            start_positions.append(0)
            end_positions.append(0)

        else:
            # Find start token
            token_start = context_start

            while (
                token_start <= context_end
                and offsets[token_start][1] <= start_char
            ):
                token_start += 1

            # Find end token
            token_end = context_end

            while (
                token_end >= context_start
                and offsets[token_end][0] >= end_char
            ):
                token_end -= 1

            start_positions.append(token_start)
            end_positions.append(token_end)

    tokenized["start_positions"] = start_positions
    tokenized["end_positions"] = end_positions

    return tokenized
```

```python
tokenized_train = train_dataset.map(
    tokenize_qa,
    batched=True,
    remove_columns=train_dataset.column_names
)
```

```python
tokenized_validation = validation_dataset.map(
    tokenize_qa,
    batched=True,
    remove_columns=validation_dataset.column_names
)
```

```python
import numpy as np

def compute_metrics(eval_pred):
    predictions, labels = eval_pred

    start_logits, end_logits = predictions
    start_labels, end_labels = labels

    start_preds = np.argmax(start_logits, axis=-1)
    end_preds = np.argmax(end_logits, axis=-1)

    start_accuracy = np.mean(start_preds == start_labels)
    end_accuracy = np.mean(end_preds == end_labels)

    exact_span_accuracy = np.mean(
        (start_preds == start_labels) &
        (end_preds == end_labels)
    )

    return {
        "start_accuracy": start_accuracy,
        "end_accuracy": end_accuracy,
        "span_accuracy": exact_span_accuracy,
    }
```

```python
from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir="./arabert-qa",
    eval_strategy="epoch",
    learning_rate=3e-5,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=3,
    weight_decay=0.01,
    logging_steps=100,
    save_strategy="epoch",
    fp16=True,
    push_to_hub=True,
    report_to="wandb",
)
```

```python
from transformers import Trainer

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train,
    eval_dataset=tokenized_validation,
    compute_metrics=compute_metrics,
)
```

```python
def answer_question(context, question):

    # Get model device
    device = next(model.parameters()).device

    inputs = tokenizer(
        question,
        context,
        max_length=384,
        truncation="only_second",
        stride=128,
        return_overflowing_tokens=True,
        return_offsets_mapping=True,
        padding="max_length",
        return_tensors="pt",
    )

    # Keep these on CPU
    offset_mapping = inputs.pop("offset_mapping")
    inputs.pop("overflow_to_sample_mapping")

    # Move model inputs to GPU
    inputs = {
        key: value.to(device)
        for key, value in inputs.items()
    }

    with torch.no_grad():
        outputs = model(**inputs)

    start_logits = outputs.start_logits
    end_logits = outputs.end_logits

    best_answer = ""
    best_score = float("-inf")

    for feature_idx in range(start_logits.shape[0]):

        start_logit = start_logits[feature_idx]
        end_logit = end_logits[feature_idx]

        offsets = offset_mapping[feature_idx]

        token_type_ids = inputs["token_type_ids"][feature_idx]

        context_tokens = torch.where(
            token_type_ids == 1
        )[0]

        if len(context_tokens) == 0:
            continue

        context_start = context_tokens[0].item()
        context_end = context_tokens[-1].item()

        start_candidates = torch.topk(
            start_logit[context_start:context_end + 1],
            k=min(
                20,
                context_end - context_start + 1
            )
        ).indices + context_start

        end_candidates = torch.topk(
            end_logit[context_start:context_end + 1],
            k=min(
                20,
                context_end - context_start + 1
            )
        ).indices + context_start

        for start_idx_tensor in start_candidates:

            start_idx = start_idx_tensor.item()

            for end_idx_tensor in end_candidates:

                end_idx = end_idx_tensor.item()

                if end_idx < start_idx:
                    continue

                if end_idx - start_idx + 1 > 20:
                    continue

                start_char = int(offsets[start_idx][0])
                end_char = int(offsets[end_idx][1])

                if end_char <= start_char:
                    continue

                answer = context[start_char:end_char].strip()

                if not answer:
                    continue

                score = (
                    start_logit[start_idx].item()
                    + end_logit[end_idx].item()
                )

                if score > best_score:
                    best_score = score
                    best_answer = answer

    return best_answer, best_score
```
